# import modules

```python
from pydantic import BaseModel, Field, ValidationError
from typing import Optional, List
from dotenv import load_dotenv
from langchain_groq import ChatGroq
import json

load_dotenv()
```

```python
import os
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_tavily import TavilySearch
from langchain_groq import ChatGroq
# Load environment variables from .env file
load_dotenv()

# Verify API keys are loaded
#note:not makes false be true so the condition code will run
if not os.getenv("GROQ_API_KEY"):
    raise ValueError("GROQ_API_KEY not found in environment variables")
if not os.getenv("TAVILY_API_KEY"):
    raise ValueError("TAVILY_API_KEY not found in environment variables")

print("API keys loaded successfully!")
```

# Initialize the Model

```python
model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.1
)
```

# use an external tool 

```python
search_tool = TavilySearch(
    max_results=5,
    search_depth="basic",  # Use "advanced" for more comprehensive results
    include_raw_content=False, #doesn't include full html page content
    include_images=False
)
```

# create the agent 

```python
agent = create_agent(
    model=model,
    tools=[search_tool]
)
```

# helper function to call the agent 

```python
def generate_and_print_response(agent, query):
    """
    Invoke the agent with a query and print the response.

    Args:
        agent: The LangChain agent instance
        query: The user query string
    """
    print(f"\n{'='*60}")
    print(f"Query: {query}")
    print(f"{'='*60}\n")

    # Invoke the agent with the query
    result = agent.invoke({
        "messages": [{"role": "user", "content": query}]
    })

    # Extract and print the final response
    #human_message = result["messages"][0]
    #ai_reasoning_message = result["messages"][1]
    #tool_call_message = result["messages"][2]
    #ai_final_message = result["messages"][3]
    final_message = result["messages"][-1]
    print(f"Response: {final_message.content}\n")

    return result

print("Helper function defined successfully!")
```

# try the agent 

```python
generate_and_print_response(agent, "What are the latest developments in artificial intelligence in 2026?")
```
